```python
import pandas as pd
```

```python
# Load data
df = pd.read_csv("models.csv")
```

```python
# Fix data (starting steps)
df = df.dropna(axis=1, how="all")
df.columns = df.columns.str.strip()

# {"CSV wala naam": "code wala naam"}
df = df.rename(columns={
    "ChatGpt": "ChatGPT",
    "Gemini": "Gemini",
    "Claude": "Claude",
    "LLaMA3": "LLaMA 3",
    "Mistral": "Mistral",
    "Falcon": "Falcon",
    "Perplexity": "Perplexity",
    "Grok": "GRok",
    "DeepSeek": "Deepseek", "Deepseek": "Deepseek",
    "Gemma": "Gemma"
})
```

```python
# Clean data
df = df.applymap(lambda x: str(x).strip().upper())
```

```python
GT = df["Ground Truth"]
total = len(df)
```

# Chatgpt

```python
model = df["ChatGPT"]

TP = ((GT == "TRUE") & (model == "TRUE")).sum()
TN = ((GT == "FALSE") & (model == "FALSE")).sum()
FP = ((GT == "FALSE") & (model == "TRUE")).sum()
FN = ((GT == "TRUE") & (model == "FALSE")).sum()

accuracy = (TP + TN) / total
precision = TP / (TP + FP) if (TP+FP)!=0 else 0
recall = TP / (TP + FN) if (TP+FN)!=0 else 0
bias = abs((model == "TRUE").sum() - (GT == "TRUE").sum()) / total
HR = (FP / (TP+FP)) * 100

print("ChatGPT")
print(TP, TN, FP, FN)
print("Accuracy:", accuracy)
print("Precision:", precision)
print("Recall:", recall)
print("Bias:", bias)
print("Hallucination Rate:", HR)
```

# Heatmap

```python

import matplotlib.pyplot as plt
import matplotlib.patches as patches
import numpy as np
from matplotlib.colors import LinearSegmentedColormap, Normalize

# DATA
models = [
    "ChatGPT", "Gemini", "Mistral", "Claude 3.5\nSonnet",
    "LLaMA 3", "Falcon", "Perplexity",
    "GRok", "DeepSeek", "Gemma"
]

accuracy = [0.84,0.91,0.84,0.87,0.80,0.74,0.84,0.80,0.82,0.74]
precision = [0.88,0.92,0.91,0.90,0.94,1.00,0.88,0.87,0.87,0.91]
recall = [0.91,0.97,0.87,0.93,0.77,0.64,0.91,0.85,0.89,0.70]
bias = [0.02,0.04,0.02,0.02,0.12,0.24,0.02,0.01,0.01,0.15]
hallucination = [12.00,7.84,8.69,10.00,5.12,0.00,12.00,12.76,12.24,8.10]

data = np.array([
    accuracy,
    precision,
    recall,
    bias,
    hallucination
])

metrics = [
    "Accuracy",
    "Precision",
    "Recall",
    "Bias Index",
    "Hallucination (%)"
]

# HEADER COLORS
header_colors = [
    "#3776D6", "#FF8C00", "#3CB043", "#E41E2B", "#8A5CC2",
    "#F1C40F", "#18B7C9", "#E84A91", "#996633", "#7B8794"
]

# CUSTOM COLORMAP
heat_cmap = LinearSegmentedColormap.from_list(
    "heat",
    ["#FF2020", "#FF9D2E", "#F5E663", "#7ACC47", "#008F39"]
)
# FIGURE
fig, ax = plt.subplots(figsize=(28, 18))
fig.patch.set_facecolor("#F2F2F2")
ax.set_xlim(0, 12)
ax.set_ylim(-3.2, 6.4)
ax.axis("off")

# Title
ax.text(
    6, 6.1,
    "LLM Performance Heatmap – Comprehensive Metric Analysis",
    ha="center",
    va="center",
    fontsize=34,
    fontweight="bold",
    color="#09153B"
)

# HEADER ROW
ax.add_patch(
    patches.Rectangle((0, 5), 1.4, 0.7,
                      facecolor="#ECECEC",
                      edgecolor="gray")
)
ax.text(
    0.7, 5.35, "Models",
    ha="center", va="center",
    fontsize=20, fontweight="bold"
)
for i, model in enumerate(models):
    x = i + 1.4

    ax.add_patch(
        patches.Rectangle(
            (x, 5), 1, 0.7,
            facecolor=header_colors[i],
            edgecolor="white",
            linewidth=2
        )
    )

    txt = "black" if model == "Falcon" else "white"

    ax.text(
        x + 0.5, 5.35,
        model,
        ha="center",
        va="center",
        fontsize=17,
        fontweight="bold",
        color=txt
    )
# HEATMAP
higher_better = [0, 1, 2]

for r in range(len(metrics)):
    y = 4 - r

    # Metric Label
    ax.add_patch(
        patches.Rectangle(
            (0, y), 1.4, 1,
            facecolor="#ECECEC",
            edgecolor="gray"
        )
    )

    subtitle = "(Higher is Better)" if r in higher_better else "(Lower is Better)"

    ax.text(
        0.7, y + 0.62,
        metrics[r],
        ha="center",
        va="center",
        fontsize=17,
        fontweight="bold"
    )

    ax.text(
        0.7, y + 0.28,
        subtitle,
        ha="center",
        va="center",
        fontsize=11
    )

    row = data[r]
    norm = Normalize(vmin=row.min(), vmax=row.max())

    for c, val in enumerate(row):
        x = c + 1.4

        score = norm(val)
        if r not in higher_better:
            score = 1 - score

        color = heat_cmap(score)

        ax.add_patch(
            patches.Rectangle(
                (x, y), 1, 1,
                facecolor=color,
                edgecolor="white",
                linewidth=1.5
            )
        )

        txt_color = "white" if score < 0.3 or score > 0.7 else "black"

        label = f"{val:.2f}% " if r == 4 else f"{val:.2f}"

        ax.text(
            x + 0.5, y + 0.5,
            label,
            ha="center",
            va="center",
            fontsize=16,
            fontweight="bold",
            color=txt_color
        )
# OUTER BORDER
ax.add_patch(
    patches.Rectangle(
        (0, 0), 11.4, 5.7,
        fill=False,
        edgecolor="#555555",
        linewidth=2.5
    )
)
# Evaluations metrics
scale_x, scale_y = 0.15, -2.95

# Main Container
ax.add_patch(
    patches.FancyBboxPatch(
        (scale_x, scale_y),
        3.6,
        2.2,
        boxstyle="round,pad=0.06",
        facecolor="#FFFFFF",
        edgecolor="#CBD5E1",
        linewidth=2.5
    )
)

# Title
ax.text(
    scale_x + 1.8,
    scale_y + 1.85,
    "Performance Scale",
    ha="center",
    va="center",
    fontsize=20,
    fontweight="bold",
    color="#0F172A"
)

# Gradient Bar
gradient = np.linspace(0, 1, 600).reshape(1, -1)

ax.imshow(
    gradient,
    extent=(
        scale_x + 0.30,
        scale_x + 3.30,
        scale_y + 1.18,
        scale_y + 1.55
    ),
    cmap=heat_cmap,
    aspect="auto"
)

# Gradient Border
ax.add_patch(
    patches.Rectangle(
        (scale_x + 0.30, scale_y + 1.18),
        3.00,
        0.37,
        fill=False,
        edgecolor="#475569",
        linewidth=1.4
    )
)

# Scale Tick Marks
tick_positions = np.linspace(
    scale_x + 0.30,
    scale_x + 3.30,
    5
)

for tick in tick_positions:
    ax.plot(
        [tick, tick],
        [scale_y + 1.10, scale_y + 1.18],
        color="#334155",
        linewidth=1.5
    )

# Performance Labels
labels = ["Poor", "Average", "Good", "Excellent"]
label_pos = [
    scale_x + 0.42,
    scale_x + 1.25,
    scale_x + 2.10,
    scale_x + 2.95
]

label_colors = [
    "#DC2626",
    "#F59E0B",
    "#65A30D",
    "#15803D"
]

for x, lbl, clr in zip(label_pos, labels, label_colors):
    ax.text(
        x,
        scale_y + 0.80,
        lbl,
        ha="center",
        fontsize=12,
        fontweight="bold",
        color=clr
    )

# Numeric Scale
numeric_labels = ["0", "25", "50", "75", "100"]

for x, num in zip(tick_positions, numeric_labels):
    ax.text(
        x,
        scale_y + 0.47,
        num,
        ha="center",
        fontsize=10,
        color="#475569"
    )

# Low/High Labels
ax.text(
    scale_x + 0.30,
    scale_y + 0.18,
    "Low Performance",
    fontsize=12,
    fontweight="bold",
    color="#B91C1C"
)

ax.text(
    scale_x + 2.05,
    scale_y + 0.18,
    "High Performance",
    fontsize=12,
    fontweight="bold",
    color="#166534"
)

# Decorative Markers
marker_x = [
    scale_x + 0.45,
    scale_x + 1.20,
    scale_x + 2.05,
    scale_x + 2.90
]

for x, clr in zip(marker_x, label_colors):
    ax.add_patch(
        patches.Circle(
            (x, scale_y + 1.73),
            0.08,
            facecolor=clr,
            edgecolor="white",
            linewidth=2
        )
    )
# Insightful Box
insight_x, insight_y = 4.10, -2.95

ax.add_patch(
    patches.FancyBboxPatch(
        (insight_x, insight_y), 5.2, 2.4,
        boxstyle="round,pad=0.04",
        facecolor="#EEF5FF",
        edgecolor="#AABBDD"
    )
)

ax.text(
    insight_x + 2.6,
    insight_y + 2.05,
    "Key Insights",
    ha="center",
    fontsize=20,
    fontweight="bold",
    color="#1E3A8A"
)

insights = [
    "✓ Best Overall Performer: Gemini",
    "🏆 Highest Precision: Falcon (1.00)",
    "⚠ Lowest Recall: Falcon (0.64)",
    "⚖ Highest Bias: Falcon (0.24)",
    "🚨 Highest Hallucination: GRok (12.76%)",
    "⭐ Lowest Hallucination: Falcon (0.00%)"
]

for i, text in enumerate(insights):
    ax.text(
        insight_x + 0.25,
        insight_y + 1.65 - i * 0.32,
        text,
        fontsize=15,
        va="center"
    )

# MODEL RANKING BOX
rank_x, rank_y = 8.65, -2.95

ax.add_patch(
    patches.FancyBboxPatch(
        (rank_x, rank_y), 3.5, 2.4,
        boxstyle="round,pad=0.04",
        facecolor="#FFF5F7",
        edgecolor="#E0B4C5"
    )
)

ax.text(
    rank_x + 1.75,
    rank_y + 2.05,
    "Model Ranking\n(Overall Performance)",
    ha="center",
    fontsize=18,
    fontweight="bold",
    color="#B91C1C"
)

ranking = [
    "1. Gemini",
    "2. Claude 3.5 Sonnet",
    "3. Mistral",
    "4. Perplexity",
    "5. ChatGPT",
    "6. DeepSeek",
    "7. GRok",
    "8. LLaMA 3",
    "9. Gemma",
    "10. Falcon"
]

for i, rank in enumerate(ranking):
    ax.text(
        rank_x + 0.3,
        rank_y + 1.7 - i * 0.17,
        rank,
        fontsize=13
    )

plt.tight_layout()
plt.show()



```
